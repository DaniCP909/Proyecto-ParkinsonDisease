```python
import os
import csv
```

```python
directorio = "/media/dcorredor/SanDisk_64/csvs/8"

for archivo in os.listdir(directorio):
    if archivo.lower().endswith(".csv"):
        ruta_csv = os.path.join(directorio, archivo)

        with open(ruta_csv, newline="", encoding="utf-8") as f:
            lector = csv.reader(f)
            next(lector)  # saltar header
            fila = next(lector, None)  # leer la única fila de datos

            if fila is not None:
                print(f"{archivo}: {fila}")
```

```python
directorio = "/media/dcorredor/SanDisk_64/csvs/8"

salida_csv = "confianzas8.csv"

confianzas = []

for archivo in sorted(os.listdir(directorio)):
    if archivo.lower().endswith(".csv"):
        ruta_csv = os.path.join(directorio, archivo)

        with open(ruta_csv, newline="", encoding="utf-8") as f:
            lector = csv.reader(f)
            next(lector)              # saltar header
            fila = next(lector, None) # leer datos

            if fila is not None:
                confianza = fila[-1]  # última columna
                confianzas.append([confianza])

# escribir CSV final
with open(salida_csv, "w", newline="", encoding="utf-8") as f:
    escritor = csv.writer(f)
    escritor.writerow(["confidence"])
    escritor.writerows(confianzas)

print(f"Archivo generado: {salida_csv}")
```
