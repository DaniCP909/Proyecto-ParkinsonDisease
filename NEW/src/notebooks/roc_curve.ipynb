```python
import pandas as pd
from sklearn.metrics import roc_curve, roc_auc_score
from matplotlib import pyplot as plt
```

```python
df_task37 = pd.read_csv("../analysis/37/tasks_37_CnnOnly_validate_2026-01-22 01:36:12.411571.csv")
df_task246 = pd.read_csv("../analysis/246/tasks_246_CnnOnly_validate_2026-01-26 14:53:44.294745.csv")
df_task346 = pd.read_csv("../analysis/346/tasks_346_CnnOnly_validate_2026-01-21 20:49:06.717545.csv")
df_task3456 = pd.read_csv("../analysis/3456/tasks_3456_CnnOnly_validate_2026-01-22 00:15:23.717703.csv")
df_task4567 = pd.read_csv("../analysis/4567/tasks_4567_CnnOnly_validate_2026-01-22 13:08:50.909797.csv")
df_task24567 = pd.read_csv("../analysis/24567/tasks_24567_CnnOnly_validate_2026-01-22 13:45:35.593277.csv")
df_task345678 = pd.read_csv("../analysis/345678/tasks_345678_CnnOnly_validate_2026-01-23 22:11:15.910425.csv")
df_task2345678 = pd.read_csv("../analysis/2345678/tasks_2345678_CnnOnly_validate_2026-01-23 23:01:12.702975.csv")

dataframes = {
    0: df_task37,
    1: df_task246,
    2: df_task346,
    3: df_task3456,
    4: df_task4567,
    5: df_task24567,
    6: df_task345678,
    7: df_task2345678,
}
```

```python
for i in range(0,8):
    df = dataframes[i]
    y_true = df["target"]
    y_score = df["park_neur confidence"]

    fpr, tpr, thresholds = roc_curve(y_true, y_score)
    auc = roc_auc_score(y_true, y_score)

    plt.figure()
    plt.plot(fpr, tpr, label=f"AUC = {auc:.3f}")
    plt.plot([0,1], [0,1], "k--")
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.title(f"Curva ROC task {i}")
    plt.legend(loc="lower right")
    plt.savefig(f"MULTITASK_ROC{i}.png", dpi=300, bbox_inches="tight")
    plt.close()
```
